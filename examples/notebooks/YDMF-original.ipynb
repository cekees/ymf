YDMF schema using strictyaml 

````python
from strictyaml import Int, Str, Seq,Map,Enum, Optional, Float, Any
DataItem = Map({"Dimensions": Int() | Seq(Int()),
                "Data": Str() | Seq(Seq(Str())),
                Optional("ItemType", default="Uniform") : Enum(["Uniform", "HyperSlab", "Function"]),                 
                Optional("Name",drop_if_none=True) : Str(),                
                Optional("DataType", default="Float") : Enum(["Char", "UChar", "Float", "Int", "UInt"]),
                Optional("Precision", default="4") : Enum(["1", "2", "4", "8"]),
                Optional("Endian", default="Native") : Enum(["Big", "Little", "Native"]),
                Optional("Format", default="YML") : Enum(["YML","HDF","Binary","TIFF"])})
Topology = Map({"Type": Enum(["Triangle", "Tetrahedron","Hexahedron"]),
                "NumberOfElements": Int(),
                "DataItem": DataItem})
Geometry = Map({"Type": Enum(["XYZ"]),
                "DataItem": DataItem})
Grid = Map({Optional("Name", drop_if_none=True): Str(),
            "Time": Float(),
            "Topology": Topology,
            "Geometry": Geometry})
TimeCollection = Map({Optional("Name", drop_if_none=True): Str(),
                      "Data": Seq(Grid)})
SpatialCollection = Map({Optional("Name", drop_if_none=True): Str(),
                         "Data": Seq(Grid)})
Domain = Map({Optional("Name", drop_if_none=True): Str(),
              Optional("TimeCollection", drop_if_none=True): TimeCollection})
ModelDomain = Map({Optional("Name", drop_if_none=True): Str(),
              Optional("Grid", drop_if_none=True): Grid})
Model = Map({"Equation": Str(),
             "BoundaryConditions": Str(),
             Optional("Name", drop_if_none=True): Str()})
YDMF = Map({"Model": Model,# | Seq(Model),
            "ModelDomain": ModelDomain,
            "Domain" : Domain})
````

Parse yaml and validate ydmf

````python
from strictyaml import dirty_load
poisson_yaml = dirty_load(poisson_data, allow_flow_style=True, schema=YDMF)
#poisson_yaml = load(poisson_data, schema=YDMF)
#doesn't allow flow style sequences so would require e.g.
#Dimensions:
#- 118084
#- 3
````

Convert ydmf to xdmf

````python
from xml.etree.ElementTree import ElementTree, Element, SubElement
#just to make it easier to check the xml
def indentXML(elem, level=0):
    i = "\n" + level*"  "
    if len(elem):
        if not elem.text or not elem.text.strip():
            elem.text = i + "  "
        for e in elem:
            indentXML(e, level+1)
            if not e.tail or not e.tail.strip():
                e.tail = i + "  "
        if not e.tail or not e.tail.strip():
            e.tail = i
    else:
        if level and (not elem.tail or not elem.tail.strip()):
            elem.tail = i
````

````python
xmlFile=open("poisson.xmf","wb")
xmlHeader = b"<?xml version=\"1.0\" ?>\n<!DOCTYPE Xdmf SYSTEM \"Xdmf.dtd\" []>\n"
xmlFile.write(xmlHeader)
tree=ElementTree(
    Element("Xdmf",
            {"Version":"2.0",
             "xmlns:xi":"http://www.w3.org/2001/XInclude"}))
for k, v in poisson_yaml.data.items():
    if k == 'Domain':
        domain = SubElement(tree.getroot(),"Domain")
    for kd, vd in v.items():
        if kd == 'TimeCollection':
            timeCollection = SubElement(domain, 
                                        "Grid",
                                        {"Name": vd['Name'],
                                         "GridType": "Collection",
                                         "CollectionType": "Temporal"})
            for gi, g in enumerate(vd['Data']):
                grid = SubElement(timeCollection,
                                  "Grid",
                                  {"GridType": "Uniform"})
                time = SubElement(grid,
                                  "Time",
                                  {"Value": str(g['Time']),
                                   "Name": str(gi)})
                t = g['Topology']
                topology = SubElement(grid,
                                      "Topology",
                                      {"Type" : t['Type'],
                                      "NumberOfElements": str(t['NumberOfElements'])})
                d = t['DataItem']
                dims = [str(dim) for dim in d['Dimensions']]
                dataItem = SubElement(topology,
                                      "DataItem",
                                      {'Format': d['Format'],
                                       'DataType': d['DataType'],
                                       'Precision': d['Precision'],
                                       'Dimensions': " ".join(dims)})
                dataItem.text = d['Data']
                ge = g['Geometry']
                geometry = SubElement(grid,
                                      "Geometry",
                                      {"Type" : ge['Type']})
                d = ge['DataItem']
                dims = [str(dim) for dim in d['Dimensions']]
                dataItem = SubElement(geometry,
                                      "DataItem",
                                      {'Format': d['Format'],
                                       'DataType': d['DataType'],
                                       'Precision': d['Precision'],
                                       'Dimensions': " ".join(dims)})
                dataItem.text = d['Data']
indentXML(tree.getroot())
tree.write(xmlFile,encoding="utf-8")
````
